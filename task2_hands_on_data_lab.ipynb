```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```
df = pd.read_csv('../data/sales_data.csv')
df.head()
```

```
print('Shape:', df.shape)
print(df.dtypes)
print(df.isnull().sum())
```

```
df['Date']=pd.to_datetime(df['Date'])
df['Month']=df['Date'].dt.to_period('M').astype(str)
category_sales=df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
region_sales=df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
category_sales, region_sales
```

```
summary={'Total Sales':df['Sales'].sum(),'Total Profit':df['Profit'].sum(),'Total Units':df['Quantity'].sum(),'Average Order Value':df['Sales'].mean()}
summary
```

```
df[['Quantity','Unit_Price','Sales','Profit']].describe()
```

```
category_sales.sort_values().plot(kind='barh',title='Sales by Category'); plt.tight_layout(); plt.show()
```

```
monthly_sales=df.groupby('Month')['Sales'].sum(); monthly_sales.plot(kind='line',marker='o',title='Monthly Sales Trend'); plt.xticks(rotation=20); plt.tight_layout(); plt.show()
```

```
df['Sales'].plot(kind='hist',bins=10,title='Distribution of Order Sales'); plt.tight_layout(); plt.show()
```
